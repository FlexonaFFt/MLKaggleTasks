```python
"""Build ff04_no_export_exact1_safe985_tail1500 attack.py from base64."""
import base64
import sys
from pathlib import Path

sys.argv = [sys.argv[0]]
WORKING_DIR = Path("/kaggle/working")
ATTACK_PY_B64 = ""

sdk_roots = sorted(Path("/kaggle/input").glob("**/aicomp_sdk"))
if sdk_roots:
    sdk_parent = str(sdk_roots[0].parent)
    if sdk_parent not in sys.path:
        sys.path.insert(0, sdk_parent)

attack_bytes = base64.b64decode(ATTACK_PY_B64)
attack_path = WORKING_DIR / "attack.py"
attack_path.write_bytes(attack_bytes)
compile(attack_bytes, str(attack_path), "exec")
print(f"attack.py written: {len(attack_bytes)} bytes")
print(f"Attack module ready: {attack_path}")
```
