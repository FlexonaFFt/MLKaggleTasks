```python
import pandas as pd
from pathlib import Path
import ast

DATA_DIR = Path('../datasets')
train_path = DATA_DIR / 'train.csv'
test_path = DATA_DIR / 'test.csv'

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)

print('train shape:', train.shape)
print('test shape:', test.shape)
```

```python
target_cols = ['winner_model_a', 'winner_model_b', 'winner_tie']
train[target_cols].sum()
```

```python
def parse_list(s):
    try:
        return ast.literal_eval(s) if isinstance(s, str) else s
    except Exception:
        return []

for col in ['prompt', 'response_a', 'response_b']:
    train[col + '_n_turns'] = train[col].apply(lambda x: len(parse_list(x)))
    train[col + '_n_chars'] = train[col].apply(lambda x: sum(len(t) for t in parse_list(x)))

train[[c for c in train.columns if c.endswith('_n_turns') or c.endswith('_n_chars')]].describe().T
```
