```python
'''Kaggriculture | Smart Agent Engine — Idle Worker Harvest

Standalone public-route submission.

This exact public policy keeps the profitable market opening and redirects
otherwise idle early workers to grow, harvest, and sell a temporary wheat crop.
The remaining route and reactive layers are preserved byte-for-byte.

Output: submission_pipe16_idle_workers.tar.gz
'''
```

```python
from pathlib import Path
import os

WORKDIR = Path('/kaggle/working') if Path('/kaggle/working').is_dir() else Path.cwd()
WORKDIR.mkdir(parents=True, exist_ok=True)
os.chdir(WORKDIR)
print('Build directory:', WORKDIR, flush=True)
```

```python
import base64
import gzip
import hashlib
from pathlib import Path

EXPECTED_MAIN_SHA256 = '827ddf2997fa442e80baadebc4cc91ffaeedec1ff333c1fa7d872b0f1d448181'
PAYLOADS = {'main.py': ''}
FILES = {name: gzip.decompress(base64.b64decode(value)) for name, value in PAYLOADS.items()}
assert hashlib.sha256(FILES['main.py']).hexdigest() == EXPECTED_MAIN_SHA256
MAIN = WORKDIR / 'main.py'
for name, data in FILES.items():
    (WORKDIR / name).write_bytes(data)
print('Wrote verified agent:', MAIN.name, len(FILES['main.py']), 'bytes', flush=True)
```

```python
import ast

source_bytes = MAIN.read_bytes()
assert hashlib.sha256(source_bytes).hexdigest() == EXPECTED_MAIN_SHA256
compile(source_bytes, 'main.py', 'exec')
tree = ast.parse(source_bytes)
assert any(isinstance(node, ast.FunctionDef) and node.name == 'agent' for node in ast.walk(tree))
assert any(isinstance(node, ast.FunctionDef) and node.name == 'agent' for node in ast.walk(tree))
print('PASS: source hash, Python syntax and agent entry point.', flush=True)
```

```python
import io
import tarfile

ARCHIVE = WORKDIR / 'submission_pipe16_idle_workers.tar.gz'
with ARCHIVE.open('wb') as raw:
    with gzip.GzipFile(filename='', mode='wb', fileobj=raw, mtime=0) as zipped:
        with tarfile.open(fileobj=zipped, mode='w', format=tarfile.GNU_FORMAT) as tar:
            for name in sorted(FILES):
                data = FILES[name]
                info = tarfile.TarInfo(name)
                info.size, info.mode, info.mtime = len(data), 0o644, 0
                tar.addfile(info, io.BytesIO(data))
with tarfile.open(ARCHIVE) as tar:
    assert tar.getnames() == sorted(FILES)
    for name, data in FILES.items():
        assert tar.extractfile(name).read() == data
print('PACKAGE READY:', ARCHIVE.name, '|', ARCHIVE.stat().st_size, 'bytes', flush=True)
print('SHA-256:', hashlib.sha256(ARCHIVE.read_bytes()).hexdigest(), flush=True)
```
