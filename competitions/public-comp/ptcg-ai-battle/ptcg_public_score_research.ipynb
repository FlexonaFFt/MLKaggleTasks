```python
# Keep Public Score visually separate from the live agent-rating system.
ax = solutions.sort_values('public_score').plot.barh(
    x='notebook', y='public_score', legend=False, color=['#9bb8c7', '#6e94a8', '#235a75']
)
ax.set_title('Top Kaggle Code cards by Public Score · snapshot 2026-10-03')
ax.set_xlabel('Public Score shown in Code listing (not live agent rating)')
ax.set_ylabel('')
ax.set_xlim(0, 600)
for bar in ax.patches:
    ax.annotate(f'{bar.get_width():.1f}', (bar.get_width(), bar.get_y() + bar.get_height()/2),
                xytext=(6, 0), textcoords='offset points', va='center')
plt.tight_layout()
plt.show()
```
