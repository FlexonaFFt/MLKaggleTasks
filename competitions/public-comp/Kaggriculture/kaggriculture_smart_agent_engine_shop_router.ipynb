```python
'''Kaggriculture | Smart Agent Engine — Shop Router

This research submission preserves an exact public shop-state router. It replays
four complete action tapes and switches only at compatible dawn checkpoints using
visible shop, market and own-farm state. The agent needs four files in its archive:
main.py, observation.py, model.json and actions.json. No external data or network
is used at runtime. Run the next cell to materialize submission.tar.gz for Kaggle.
'''
```

```python
'''Materialize the complete standalone Shop Router archive.

The action tapes and policy model are intentionally kept byte-for-byte from the
source solution. The archive must retain every listed file, not only main.py.
'''
import base64
import hashlib
import tarfile
from pathlib import Path

ARCHIVE_B64 = ''
archive = Path("submission.tar.gz")
archive.write_bytes(base64.b64decode(ARCHIVE_B64))

'''Verify the submission boundary before it is selected in Kaggle.'''
with tarfile.open(archive, "r:gz") as tar:
    expected = ["main.py", "observation.py", "model.json", "actions.json"]
    assert tar.getnames() == expected
    main = tar.extractfile("main.py").read().decode()
    compile(main, "main.py", "exec")

print("variant: smart_agent_engine_shop_router_exact")
print("archive SHA-256:", hashlib.sha256(archive.read_bytes()).hexdigest())
print("submission.tar.gz: PASS")
```
