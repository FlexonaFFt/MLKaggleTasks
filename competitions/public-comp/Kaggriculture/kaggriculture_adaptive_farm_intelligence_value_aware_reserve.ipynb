```python
import base64, hashlib, tarfile, io
from pathlib import Path

VARIANT = "adaptive_farm_intelligence_value_aware_reserve_exact"
EXPECTED_MAIN_SHA256 = "2b97e2c653018ac4aeffb8463ec91c8f26b097b4cef81289acc25ccdbc68f916"
MAIN_B64 = ""

main_bytes = base64.b64decode(MAIN_B64)
assert hashlib.sha256(main_bytes).hexdigest() == EXPECTED_MAIN_SHA256
Path("main.py").write_bytes(main_bytes)
compile(Path("main.py").read_text(), "main.py", "exec")

archive = Path("submission.tar.gz")
with tarfile.open(archive, "w:gz") as tf:
    info = tarfile.TarInfo("main.py")
    info.size = len(main_bytes)
    info.mtime = 0
    tf.addfile(info, io.BytesIO(main_bytes))
with tarfile.open(archive, "r:gz") as tf:
    assert tf.getnames() == ["main.py"]
    assert tf.extractfile("main.py").read() == main_bytes

print("variant:", VARIANT)
print("main.py SHA-256:", EXPECTED_MAIN_SHA256)
print("syntax: PASS")
print("archive structure: PASS")
```
